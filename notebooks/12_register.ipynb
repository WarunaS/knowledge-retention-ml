# Stages 11-12 - Track and register

**Track:** every run folder has `run_info.json` (git commit, uncommitted-changes flag, library versions, data fingerprints). The index below lists all runs.

**Register:** the model is chosen on **validation**; registration is refused unless the run was evaluated on the gold set (Stage 9). Each version gets a model card. Batch inference (Stage 14) and the transfer test (X1) load the current version.
